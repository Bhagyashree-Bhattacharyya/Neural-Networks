```python
def image2vector(image):
    vector = image.reshape(image.shape[0]*image.shape[1]*image.shape[2],1)
    return vector
```

```python
def normalize_rows(x):
    """
    Implement a function that normalizes each row of the matrix x (to have unit length).

    Argument:
    x -- A numpy matrix of shape (n, m)

    Returns:
    x -- The normalized (by row) numpy matrix. You are allowed to modify x.
    """
    x_norm = np.linalg.norm(x,keepdims=True, axis=1)
    x = x/x_norm

    return x
```

```python
x = np.array([[0, 3, 4],
              [1, 6, 4]])
print(normalize_rows(x))
```

```python
def softmax(x):
    x_exp = np.exp(x)
    x_sum = np.sum(x_exp,axis=1,keepdims=True)
    s = x_exp/x_sum
    return s
```

```python
t_x = np.array([[9, 2, 5, 0, 0],
                [7, 5, 0, 0 ,0]])
print(softmax(t_x))
```

```python
def L1(yhat, y):
    loss = sum(abs(y-yhat))
    return loss
```

```python
yhat = np.array([.9, 0.2, 0.1, .4, .9])
y = np.array([1, 0, 0, 1, 1])

print("L1 loss (MAE) = " + str(L1(yhat, y)))
```

```python
def L2(yhat, y):
    x = yhat-y
    loss =  np.dot(x,x)
    return loss
```

```python
yhat = np.array([.9, 0.2, 0.1, .4, .9])
y = np.array([1, 0, 0, 1, 1])

print("L2 loss (MSE)= " + str(L2(yhat, y)))
```
